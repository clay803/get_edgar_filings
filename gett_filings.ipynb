## Public Financials Retreival

** Steps:
1. Input company ticker and review company facts.
2. Get list of 10ks and 10Qs.
3. Get dataframes of the three financials statments for the input time horizon.
4. Export results to excel to be modeled.

```python
from edgar import *
from edgar.xbrl import XBRL
import pandas as pd

# Need to set your identity to use EDGAR API; input your email address
set_identity('purblindstocks@gmail.com')
```

```python
# Set company ticker
company = Company('LOVE')
if company.not_found:
    print("Company not found")
company.docs
```

```python
# Get company facts (my primary purpose here is to see their fiscal years and how many periods of financials we have available)
facts = company.get_facts()
facts
```

```python
# Set the filing that you want to retrieve. In this case, we are getting 10 (or max) 10-K filings which contain the financial statements.
filings = company.get_filings(form="10-K").head(10)
filings
```

```python
# Extract the financial statements from the filings
multi_financials = MultiFinancials.extract(filings)

# Assign the Income Statement, Balance Sheet, and Cash Flow Statement as objects
balance_sheet = multi_financials.balance_sheet()
income_statement = multi_financials.income_statement()
cashflow_statement = multi_financials.cashflow_statement()
```

```python
# Print the object types from above
print(type(multi_financials))
print(type(balance_sheet))
print(type(income_statement))
print(type(cashflow_statement))
```

```python
# Print the multi-year income statement object
print("Multi-Year Income Statement")
print(income_statement)
```

```python
# Print the multi-year balance sheet object
print("Multi-Year Balance Sheet")
print(balance_sheet)
```

```python
# Print the multi-year cashflow statement object
print("Multi-Year Cash Flow Statement")
print(cashflow_statement)
```

```python
# Create lists of the column names within each financial statement
is_df = income_statement.to_dataframe()
is_all_columns = is_df.columns.tolist()

bs_df = balance_sheet.to_dataframe()
bs_all_columns = bs_df.columns.tolist()

cfs_df = cashflow_statement.to_dataframe()
cfs_all_columns = cfs_df.columns.tolist()

is_all_columns, bs_all_columns, cfs_all_columns
```

```python
# Remove the 'concept' column from each list
is_columns = is_all_columns
is_columns.pop(1)
bs_columns = bs_all_columns
bs_columns.pop(1)
cfs_columns = cfs_all_columns
cfs_columns.pop(1)

# Check to see if they are gone
is_columns, bs_columns, cfs_columns
```

```python
# Reset the financial statement dataframes without the 'concept' column
income_statement_df = is_df[is_columns]
balance_sheet_df = bs_df[bs_columns]
cashflow_statement_df = cfs_df[cfs_columns]

income_statement_df, balance_sheet_df, cashflow_statement_df
```

```python
# Check the data types of each dataframe
print(income_statement_df.dtypes)
print(balance_sheet_df.dtypes)
print(cashflow_statement_df.dtypes)
```

```python
# Since the "LOVE" example has a balance sheet full of objects, rather than integers or floats, we need to convert them. Same thing with half of the cash flow statement.
# Creating a function to optimize the dataframes by converting the columns to numeric types

def optimized_financial_df(df, cols):
    """" Convert specified columns of a DataFrame to numeric, coercing errors to NaN."""
    optimized_df = df.copy()
    
    # Get the number of columns to determine how many need to be converted
    num_cols = len(optimized_df.columns)
    
    # Convert all columns exept the first one (label) to numeric
    for i in range(1, num_cols):
        optimized_df.iloc[:, i] = pd.to_numeric(optimized_df.iloc[:, i], errors="coerce")
        # Note: errors="coerce" will convert non-numeric values to NaN, and not raise an error so that the loop can continue
        
    # Create dtype dictionary for the specified columns
    dtype_dict = {}
    for i in range(1, num_cols):
        dtype_dict[optimized_df.columns[i]] = 'float64'
        
    # Apply the dtype conversion
    optimized_df = optimized_df.astype(dtype_dict)
    
    return optimized_df
      
    
```

```python
# Applying the above functions to our intended financial statement dataframes

income_statement_opt = optimized_financial_df(income_statement_df, is_columns)
balance_sheet_opt = optimized_financial_df(balance_sheet_df, bs_columns)
cash_flow_opt = optimized_financial_df(cashflow_statement_df, cfs_columns)

print(income_statement_opt.info())
print(balance_sheet_opt.info())
print(cash_flow_opt.info())
```

```python
# Export to excel

with pd.ExcelWriter("Financial Statements.xlsx", engine="openpyxl") as writer:
    income_statement_opt.to_excel(writer, index=False, header=True, sheet_name="Income Statement")
    balance_sheet_opt.to_excel(writer, index=False, header=True, sheet_name="Balance Sheet")
    cash_flow_opt.to_excel(writer, index=False, header=True, sheet_name="Cash Flow")     
```
